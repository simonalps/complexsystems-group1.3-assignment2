# Forest Fire Model Replication

```python
# imports
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
import imageio_ffmpeg

from matplotlib.animation import FuncAnimation
from matplotlib.colors import ListedColormap
from forest_fire import model_step, run_model

mpl.rcParams["animation.ffmpeg_path"] = imageio_ffmpeg.get_ffmpeg_exe()
```

```python
# function used to run the model and create the videos
def animate(ratio):
    f = p / ratio
    L = np.zeros((n, n))
    rng = np.random.default_rng(seed)

    #burn-in --> allowing the forest to fill up, these timesteps are not included in the video
    #feel free to change the number of time-steps as you desire
    L = run_model(L, n, p, f, 400, rng)

    fig, ax = plt.subplots()
    img = ax.imshow(
        L,
        cmap=cmap,
        vmin=0,
        vmax=2
    )

    ax.set_title(f'p/f = {ratio}')
    ax.set_xticks([])
    ax.set_yticks([])


    def update(frame):
        global L
        L = model_step(L, n, p, f, rng)
        img.set_data(L)
        return [img]


    ani = FuncAnimation(
        fig,
        update,
        frames=frames,
        interval=1000 / fps,
        blit=False,
        init_func=lambda: [img]
    )

    plt.close(fig)
    ani.save(f'forest_fire_ratio_{ratio}.mp4', writer="ffmpeg", fps=fps)
```

```python
# Taking care of model parameters before running
n = 100     # n x n forest grid
p = 1.0e-3      # P(growth)
seed = 123
```

```python
# Video parameters
fps = 20
duration = 45
frames = fps * duration
cmap = ListedColormap(["black", "green", "red"])
```

For the purpose of presenting, we ran the model with these ratios to observe system behaviour. Feel free to change the ratio as you desire. The presentation included the animations with `ratio` = 10 and `ratio` = 80 (with a small tweak to start with an empty forest, running this code will still include the burn-in steps!).

```python
animate(10)
```
