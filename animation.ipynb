# Forest Fire Model Replication

```python
# imports
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
import imageio_ffmpeg

from matplotlib.animation import FuncAnimation
from matplotlib.colors import ListedColormap
from forest_fire import model_step, run_model

mpl.rcParams["animation.ffmpeg_path"] = imageio_ffmpeg.get_ffmpeg_exe()
```

```python
def animate(ratio):
    f = p / ratio
    L = np.zeros((n, n))
    rng = np.random.default_rng(seed)

    #burn-in
    L = run_model(L, n, p, f, 400, rng)

    fig, ax = plt.subplots()
    img = ax.imshow(
        L,
        cmap=cmap,
        vmin=0,
        vmax=2
    )

    ax.set_title(f'Forest Fire Model - p/f = {ratio}')
    ax.set_xticks([])
    ax.set_yticks([])


    def update(frame):
        global L

        L = model_step(L, n, p, f, rng)

        img.set_data(L)

        return [img]


    ani = FuncAnimation(
        fig,
        update,
        frames=frames,
        interval=1000 / fps,
        blit=False,
        init_func=lambda: [img]
    )

    plt.close(fig)

    ani.save(f'forest_fire_ratio_{ratio}.mp4', writer="ffmpeg", fps=fps)
```

```python
# Taking care of parameters before running
n = 100
p = 1.0e-3
seed = 42
```

```python
fps = 20
duration = 45

frames = fps * duration

cmap = ListedColormap(["black", "green", "red"])
```

```python
animate(10)
```

```python
animate(100)
```

```python
animate(70)
```

```python
f = p / 10
L = np.zeros((n, n))
rng = np.random.default_rng(seed)

fig, ax = plt.subplots()
img = ax.imshow(
    L,
    cmap=cmap,
    vmin=0,
    vmax=2
)

ax.set_title(f'Forest Fire Model - tipping')
ax.set_xticks([])
ax.set_yticks([])


def update(frame):
    global L

    L = model_step(L, n, p, f, rng)

    img.set_data(L)

    return [img]


ani = FuncAnimation(
    fig,
    update,
    frames=frames,
    interval=1000 / fps,
    blit=False,
    init_func=lambda: [img]
)

plt.close(fig)

ani.save(f'forest_fire_ratio_tipping.mp4', writer="ffmpeg", fps=fps)
```
